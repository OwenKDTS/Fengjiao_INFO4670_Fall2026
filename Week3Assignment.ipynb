```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
NAVY = "#22303A"
RUST = "#C0562F"
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"]   = (8, 5)
plt.rcParams["axes.titlesize"]   = 13
plt.rcParams["axes.titleweight"] = "bold"
```

```python
#I am choosing to use the local upload method and not link my google drive
DATA_DIR = "."
import os
students = pd.read_csv(os.path.join(DATA_DIR, "student_records.csv"))
weekly   = pd.read_csv(os.path.join(DATA_DIR, "weekly_activity.csv"))
courses = pd.read_csv(os.path.join(DATA_DIR, "course_enrollments.csv"))

#students.head()
```

```python
students.head()
```

```python
# weekly work hours
fig, ax = plt.subplots()
sns.histplot(data=students, x="work_hours_per_week", color=NAVY, ax=ax)
ax.set_title("Distribution of weekly work hours for students")
ax.set_xlabel("Work hours (per week)")
ax.set_ylabel("Number of students")
plt.show()
```

Scope: This is a histogram showing how many hours students work each week.
Track: The goal is to see how much students are working outside of class.
Articulate: Most students work around 15 to 20 hours a week, but there is a small group working 40 or more hours.
Respond: We should check if the students working full-time need more flexible or online class options.

```python
# enrollment by major
order = students["major"].value_counts().index
fig, ax = plt.subplots()
sns.countplot(data=students, y="major", order=order, color=NAVY, ax=ax)
ax.set_title("Student enrollment by major")
ax.set_xlabel("number of students")
ax.set_ylabel("major")
plt.show()
```

Scope: This is a bar chart counting the number of students in each major.
Track: The goal is to see which majors have the most students.
Articulate: Cyber Security and Data Science have the most students, while Health Informatics is the smallest.
Respond: We need to make sure the biggest majors have enough advisors to handle the crowd.

```python
# gpa by major
fig, ax = plt.subplots(figsize=(9, 5))
sns.boxplot(data=students, x="final_gpa", y="major", color=NAVY, ax=ax)
ax.set_title("GPA distribution across majors")
ax.set_xlabel("Final GPA")
ax.set_ylabel("Major")
plt.show()
```

Scope: This is a box plot comparing final GPAs across different majors. Track: The goal is to see if grades change depending on what a student studies. Articulate: Average grades are around 3.0 across all majors. However, every major has a group of students sitting at exactly a 0.00 GPA. Respond: We need to figure out why so many students have a flat zero and if they dropped out without unenrolling.

```python
# commute vs gpa
fig, ax = plt.subplots()
sns.regplot(data=students, x="commute_miles", y="final_gpa",
            scatter_kws={"s": 12, "color": NAVY, "alpha": 0.4},
            line_kws={"color": RUST}, ax=ax)
ax.set_title("Impact of commute distance on final GPA")
ax.set_xlabel("Commute distance (miles)")
ax.set_ylabel("Final GPA")
plt.show()
```

Scope: This is a scatterplot comparing how far students commute to their final GPA. Track: The goal is to see if a longer commute hurts a student's grades. Articulate: Grades drop just a little bit as the commute gets longer, and that stripe of 0.00 GPAs is still there no matter the distance. Respond: We should see if commuters need better access to online tutoring, but the drive itself might not be the direct cause of the lower grades.

```python
# emgagement over semester
weekly_avg = weekly.groupby("week")["minutes_active"].mean().reset_index()

fig, ax = plt.subplots()
sns.lineplot(data=weekly_avg, x="week", y="minutes_active",
             color=NAVY, marker="o", ax=ax)
ax.set_title("Average weekly engagement scross the semester")
ax.set_xlabel("Week of semester")
ax.set_ylabel("Average active minutes")
plt.show()
```

Scope: This is a line chart tracking the average time students spend in the online class system each week. Track: The goal is to see when students are most active online during the semester. Articulate: Activity stays steady early on, jumps up near midterms, and then crashes hard around week 15 for Thanksgiving. Respond: We should look at individual student charts to find the specific people who stop logging in early so we can check on them.

```python
#gpa by housing type
fig, ax = plt.subplots(figsize=(9, 5))
sns.violinplot(data=students, x="final_gpa", y="housing", color=NAVY, ax=ax)
ax.set_title("Distribution of GPA by housing type")
ax.set_xlabel("Final gpa")
ax.set_ylabel("Housing type")
plt.show()
```

Scope: This is a violin plot comparing where students live to their final grades. Track: The goal is to see if living on campus helps your grades. Articulate: The bulk of grades sit around 3.0 no matter where students live, but the chart shows the housing data is very messy with separate rows for the same categories. Respond: We have to clean up and combine those messy housing labels before we try to make any real decisions based on this chart.
